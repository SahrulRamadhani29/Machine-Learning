## Data teman

read_csv

```python
import pandas as pd

data_teman = pd.read_csv('data_30_teman.csv')
data_teman.columns = [
    'Nama Lengkap', 'Jenis Kelamin', 'Kota Asal',
    'Tinggi Badan (cm)', 'Berat Badan (kg)',
    'Asal Kelas', 'Target IPS'
]
data_teman = data_teman[data_teman['Nama Lengkap'].str.strip() != 'Sahrul Ramadhani']
data_teman = data_teman.reset_index(drop=True)
data_teman
```
