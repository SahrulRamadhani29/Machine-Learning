## Perbandingan CountVectorizer dan TF IDF

## Persiapan data
Data yang dipakai adalah spam.csv. Jika file belum ada, notebook mengunduhnya otomatis.

```python
from pathlib import Path
from urllib.request import Request, urlopen

data_path = Path("spam.csv")
if not data_path.exists():
    request = Request("https://1812311909-files.gitbook.io/~/files/v0/b/gitbook-x-prod.appspot.com/o/spaces%2FYHOv2XH8FJMJSMsnhwNa%2Fuploads%2F5SOal6OzJik1AsEMRu5M%2Fspam.csv?alt=media&token=29d9bc0d-7698-4e86-9404-aa3559f1ed31", headers={"User-Agent": "Mozilla/5.0"})
    with urlopen(request) as response:
        data_path.write_bytes(response.read())
print(f"Dataset siap: {data_path.resolve()}")
```

## Load dan preprocessing data

```python
import pandas as pd
from sklearn.model_selection import train_test_split

df = pd.read_csv(data_path, encoding="latin-1").iloc[:, :2]
df.columns = ["Labels", "SMS"]
df["Labels"] = df["Labels"].map({"spam": 1, "ham": 0})
X = df["SMS"]
y = df["Labels"]
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.2, random_state=50
)
```

## Multinomial Naive Bayes dengan CountVectorizer

```python
from sklearn.feature_extraction.text import CountVectorizer, TfidfVectorizer
from sklearn.metrics import accuracy_score, classification_report, confusion_matrix
from sklearn.naive_bayes import MultinomialNB

count_vectorizer = CountVectorizer(stop_words="english")
X_train_count = count_vectorizer.fit_transform(X_train)
X_test_count = count_vectorizer.transform(X_test)
count_model = MultinomialNB()
count_model.fit(X_train_count, y_train)
count_train_pred = count_model.predict(X_train_count)
count_test_pred = count_model.predict(X_test_count)
count_train_accuracy = accuracy_score(y_train, count_train_pred)
count_test_accuracy = accuracy_score(y_test, count_test_pred)
print(f"Akurasi train CountVectorizer: {count_train_accuracy}")
print(f"Akurasi test CountVectorizer: {count_test_accuracy}")
print("Confusion Matrix CountVectorizer:")
print(confusion_matrix(y_test, count_test_pred))
print(classification_report(y_test, count_test_pred))
```

## Multinomial Naive Bayes dengan TF IDF

```python
tfidf_vectorizer = TfidfVectorizer(stop_words="english")
X_train_tfidf = tfidf_vectorizer.fit_transform(X_train)
X_test_tfidf = tfidf_vectorizer.transform(X_test)
tfidf_model = MultinomialNB()
tfidf_model.fit(X_train_tfidf, y_train)
tfidf_train_pred = tfidf_model.predict(X_train_tfidf)
tfidf_test_pred = tfidf_model.predict(X_test_tfidf)
tfidf_train_accuracy = accuracy_score(y_train, tfidf_train_pred)
tfidf_test_accuracy = accuracy_score(y_test, tfidf_test_pred)
print(f"Akurasi train TF IDF: {tfidf_train_accuracy}")
print(f"Akurasi test TF IDF: {tfidf_test_accuracy}")
print("Confusion Matrix TF IDF:")
print(confusion_matrix(y_test, tfidf_test_pred))
print(classification_report(y_test, tfidf_test_pred))
```

## Kesimpulan
Pada pembagian data yang sama, CountVectorizer memberi akurasi data uji lebih tinggi daripada TF IDF. Karena itu CountVectorizer menjadi pilihan yang lebih baik pada data spam.csv dalam percobaan ini.
