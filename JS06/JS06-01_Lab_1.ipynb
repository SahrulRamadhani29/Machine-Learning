## Langkah 1: Import library

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy import stats
import seaborn as sns
```

## Langkah 2: Membuat data dummy

```python
from sklearn.datasets import make_blobs
X, y = make_blobs(n_samples=50, centers=2,
                  random_state=0, cluster_std=0.60)
plt.scatter(X[:, 0], X[:, 1], c=y, s=50, cmap='autumn')
```

## Langkah 3: Ilustrasi garis pemisah

```python
xfit = np.linspace(-1, 3.5)
plt.scatter(X[:, 0], X[:, 1], c=y, s=50, cmap='autumn')
for m, intercept in [(1, 0.65), (0.5, 1.6), (-0.2, 2.9)]:
    plt.plot(xfit, m * xfit + intercept, '-k')
plt.plot([0.6], [2.1], 'x', color='red', markersize=10)
plt.xlim(-1, 3.5)
plt.show()
```

## Langkah 4: Ilustrasi margin

```python
xfit = np.linspace(-1, 3.5)
plt.scatter(X[:, 0], X[:, 1], c=y, s=50, cmap='autumn')

for m, b, d in [(1, 0.65, 0.33), (0.5, 1.6, 0.55), (-0.2, 2.9, 0.2)]:
    yfit = m * xfit + b
    plt.plot(xfit, yfit, '-k')
    plt.fill_between(xfit, yfit - d, yfit + d, edgecolor='none',
                     color='#AAAAAA', alpha=0.4)

plt.xlim(-1, 3.5)
```

## Langkah 5: Fitting model

```python
from sklearn.svm import SVC
model = SVC(kernel='linear', C=1E10)
model.fit(X, y)
```

## Fungsi visualisasi

```python
def plot_svc_decision_function(model, ax=None, plot_support=True):
    
    if ax is None:
        ax = plt.gca()
    xlim = ax.get_xlim()
    ylim = ax.get_ylim()

    x = np.linspace(xlim[0], xlim[1], 30)
    y = np.linspace(ylim[0], ylim[1], 30)
    Y, X = np.meshgrid(y, x)
    xy = np.vstack([X.ravel(), Y.ravel()]).T
    P = model.decision_function(xy).reshape(X.shape)

    ax.contour(X, Y, P, colors='k',
               levels=[-1, 0, 1], alpha=0.5,
               linestyles=['--', '-', '--'])

    if plot_support:
        ax.scatter(model.support_vectors_[:, 0],
                   model.support_vectors_[:, 1],
                   s=300, linewidth=1, facecolors='none', edgecolors='k');
    ax.set_xlim(xlim)
    ax.set_ylim(ylim)
```

## Batas keputusan dan support vector

```python
plt.scatter(X[:, 0], X[:, 1], c=y, s=50, cmap='autumn')
plot_svc_decision_function(model)
```

## Perbandingan jumlah data

```python
def plot_svm(N=10, ax=None):
    X, y = make_blobs(n_samples=200, centers=2,
                      random_state=0, cluster_std=0.60)
    X = X[:N]
    y = y[:N]
    model = SVC(kernel='linear', C=1E10)
    model.fit(X, y)
    
    ax = ax or plt.gca()
    ax.scatter(X[:, 0], X[:, 1], c=y, s=50, cmap='autumn')
    ax.set_xlim(-1, 4)
    ax.set_ylim(-1, 6)
    plot_svc_decision_function(model, ax)

fig, ax = plt.subplots(1, 2, figsize=(16, 6))
fig.subplots_adjust(left=0.0625, right=0.95, wspace=0.1)
for axi, N in zip(ax, [60, 120]):
    plot_svm(N, axi)
    axi.set_title('N = {0}'.format(N))
```

## Visualisasi interaktif

```python
from ipywidgets import interact, fixed
interact(plot_svm, N=[10, 200], ax=fixed(None))
```

## Percobaan dataset lain

Pengantar Lab meminta dataset lain. Percobaan ini memakai dua kelas Iris, yaitu setosa dan versicolor, dengan fitur panjang dan lebar petal. Iris tersedia langsung dari Scikit Learn. Jika dosen menentukan dataset tertentu, bagian ini perlu mengikuti arahan tersebut.

```python
from sklearn.datasets import load_iris
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.metrics import accuracy_score, confusion_matrix

iris = load_iris()
mask = iris.target < 2
iris_X = iris.data[mask, 2:4]
iris_y = iris.target[mask]
iris_X_train, iris_X_test, iris_y_train, iris_y_test = train_test_split(
    iris_X, iris_y, test_size=0.3, random_state=42, stratify=iris_y
)
iris_scaler = StandardScaler()
iris_X_train = iris_scaler.fit_transform(iris_X_train)
iris_X_test = iris_scaler.transform(iris_X_test)
iris_model = SVC(kernel='linear', C=1.0)
iris_model.fit(iris_X_train, iris_y_train)
iris_pred = iris_model.predict(iris_X_test)
print('Akurasi uji Iris:', accuracy_score(iris_y_test, iris_pred))
print('Confusion matrix Iris:')
print(confusion_matrix(iris_y_test, iris_pred))
plt.scatter(iris_X_train[:, 0], iris_X_train[:, 1], c=iris_y_train, cmap='autumn')
plot_svc_decision_function(iris_model)
plt.xlabel('Panjang petal setelah standardisasi')
plt.ylabel('Lebar petal setelah standardisasi')
plt.show()
```
