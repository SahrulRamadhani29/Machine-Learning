## Langkah 2: Import library

```python
import hnswlib
import numpy as np
import time
from sklearn.neighbors import NearestNeighbors
```

## Langkah 3: Membuat dataset dummy

```python
num_elements = 1000
dim = 2
data = np.random.random((num_elements, dim)).astype(np.float32)

query = np.array([[0.5, 0.5]], dtype=np.float32)
k = 5
```

## Langkah 4: Membuat model NN

```python
nn = NearestNeighbors(n_neighbors=k, algorithm='brute', metric='euclidean')
nn.fit(data)

start = time.time()
distances, indices = nn.kneighbors(query)
end = time.time()

print("=== Exact NN ===")
print("Indices:", indices)
print("Distances:", distances)
print("Waktu:", end - start, "s")
```

## Langkah 5: Membuat model HNSW

```python
p = hnswlib.Index(space='l2', dim=dim)  

p.init_index(max_elements=num_elements, ef_construction=100, M=16)

p.add_items(data)

p.set_ef(50)

start = time.time()
labels, distances = p.knn_query(query, k=k)
end = time.time()

print("\n=== HNSW ===")
print("Indices:", labels)
print("Distances:", distances)
print("Waktu:", end - start, "s")
```

Dataset pada panduan tidak memakai seed, sehingga indeks tetangga dapat berubah setiap eksekusi. Scikit Learn menampilkan jarak Euclidean, sedangkan HNSW dengan space l2 menampilkan kuadrat jaraknya.
